## 1. My lane as an ML task (type)

*Classification, clustering, ranking, or scoring — which one, and why?*

```python
Task Type: Ranking / Binary Classification

We want to rank pages based on their probability of declining or predict whether a page will decay (1) or remain stable/growing (0).
```

## 2. Target or proxy

*What would you predict? Where does that label come from — observed outcome or a defined rule?*

```python
Target / Proxy Variable: is_declining (Derived from trend_direction == 'down').

Definition: A binary flag where 1 indicates a page is in a downward traffic trend over the last 90 days, requiring refresh review, and 0 otherwise.
```

## 3. Success metric

*One metric you can defend. What number means 'good'?*

```python
Primary Metric: Precision@50

Why: Content teams have limited weekly capacity (e.g., updating 50 pages/week). We need the top 50 flagged pages to actually be declining so we don't waste writer resources on healthy pages.
```

## 4. The unit of analysis, as a real dataframe

*Load your lane's slice and show it: one row = one what?*

```python
import pandas as pd

# Load starter data
df = pd.read_csv("https://raw.githubusercontent.com/flyrank-bih/flyrank-ml-internship-starter/main/data/raw/content_refresh_anonymized.csv")

# Create the target variable
df['is_declining'] = (df['trend_direction'] == 'down').astype(int)

# Unit of Analysis: 1 Row = 1 Search Page
unit_of_analysis = df[['page_id', 'search_volume', 'impressions_90d', 'word_count', 'trend_direction', 'is_declining']]

print("Unit of Analysis: One row represents a single anonymized page")
print(f"Shape: {unit_of_analysis.shape}")
unit_of_analysis.head(5)
```

## 5. Why ML beats a fixed rule here

*What makes the pattern too messy for an if-statement?*

```python
Why ML: Hand-written rules (like "flag pages older than 365 days") reach only ~24% Precision@50 because page age alone ignores complex interactions between impressions, CTR, and position drops. ML models (like Random Forest) evaluate multi-variable feature patterns simultaneously, achieving ~74% Precision@50 on the same evaluation holdout.
```
