## 3. Quick look at the data (2-3 real numbers)

*Load the starter CSV below and show 2-3 real numbers that make your lane look worth the next 7 weeks.*

```python
import pandas as pd

# 1. Load the starter dataset
df = pd.read_csv("data/raw/content_refresh_anonymized.csv")

# Number 1: Search volume correlation with 90-day impressions
corr = df["search_volume"].corr(df["impressions_90d"])
print(f"Number 1 - Search volume vs impressions correlation: {corr:.3f}")

# Number 2: Median word count comparison between 'down' and 'up' pages
wc = df.groupby("trend_direction")["word_count"].median()
print(f"Number 2 - Median word count (down vs up): {wc['down']} vs {wc['up']}")

# Number 3: Total rows in sample
print(f"Number 3 - Total dataset sample rows: {len(df)}")
```
